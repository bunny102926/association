```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#Given transactions
data = [
    ['Bread', 'Butter', 'Jam', 'Juice', 'Milk'],
    ['Jam', 'Juice', 'Bread', 'Butter'],
    ['Juice', 'Milk', 'Bread'],
    ['Butter', 'Juice', 'Biscuit', 'Milk'],
    ['Bread', 'Milk', 'Butter']
]
```

```python
items = ['Bread','Butter','Jam','Juice','Milk','Biscuit']

df = pd.DataFrame([
    [product in transaction for product in items] for transaction in data ],
columns=items)
```

```python
print(df)
```

```python
#find frequent itemsets 
freq = apriori(df,min_support=0.4,use_colnames=True)
print("Frequent Itemsets:")
print(freq)
```

```python
# Generate association rules
rules = association_rules(
    freq, metric="confidence", min_threshold=0.5
)
```

```python
print("\nAssociation Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
